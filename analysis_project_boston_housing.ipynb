# تحليل عوامل البيئة والاجتماع في بوسطن: دراسة مقارنة بين تأثير المناطق القريبة والبعيدة عن نهر تشارلز على قيم المنازل

```python
import pandas as pd

# تحميل البيانات (تأكد من استبدال 'path_to_your_dataset.csv' بالمسار الصحيح للملف)
df = pd.read_csv('path_to_your_dataset.csv')

# استكشاف البيانات
df.info()  # معلومات حول الأعمدة
df.describe()  # إحصائيات وصفية للبيانات
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# رسم العلاقة بين NOX و PTRATIO
sns.scatterplot(x='NOX', y='PTRATIO', data=df)
plt.title('Relationship between NOX and PTRATIO')
plt.show()
```

```python
# رسم Boxplot لمقارنة MEDV بين المناطق القريبة والبعيدة عن نهر تشارلز
sns.boxplot(x='CHAS', y='MEDV', data=df)
plt.title('Comparison of Median Home Values based on Proximity to Charles River')
plt.show()
```

```python
import folium

# إنشاء خريطة تفاعلية لبوسطن
map_boston = folium.Map(location=[42.3601, -71.0589], zoom_start=12)

# إضافة Marker للموقع (يمكنك تخصيصه باستخدام بياناتك الجغرافية)
folium.Marker([42.3601, -71.0589], popup='Boston').add_to(map_boston)

# عرض الخريطة
map_boston
```

```python
from scipy.stats import ttest_ind

# تقسيم البيانات إلى مجموعتين بناءً على CHAS
chas_1 = df[df['CHAS'] == 1]['MEDV']
chas_0 = df[df['CHAS'] == 0]['MEDV']

# إجراء اختبار t
t_stat, p_value = ttest_ind(chas_1, chas_0)
print(f'T-statistic: {t_stat}, P-value: {p_value}')
```

```python
import statsmodels.api as sm

# إعداد البيانات
X = df['DIS']
Y = df['MEDV']
X = sm.add_constant(X)  # إضافة ثابت للانحدار

# إنشاء نموذج الانحدار
model = sm.OLS(Y, X).fit()

# عرض نتائج الانحدار
print(model.summary())
```
